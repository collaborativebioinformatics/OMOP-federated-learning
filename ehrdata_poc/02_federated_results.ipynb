# Federated model evaluation

Compares the NVFlare FedAvg global model against models trained on a single ICU only, and against a centralized model pooling every site's training data.

Run `python run.py`, then `01_explore.ipynb`.

```python
from pathlib import Path

import ehrapy as ep
import ehrdata as ed
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.metrics import roc_auc_score, roc_curve

from data import load_shard
from training import fit, load_global_model, risk_scores

DATA, WORKSPACE, JOB, EPOCHS = Path("data"), Path("workspace"), "physionet_fedavg", 10

shards = {p.stem: load_shard(p) for p in sorted((DATA / "shards").glob("*.npz"))}
X_train = np.vstack([s["X_train"] for s in shards.values()])
y_train = np.concatenate([s["y_train"] for s in shards.values()])
X_test = np.vstack([s["X_test"] for s in shards.values()])
y_test = np.concatenate([s["y_test"] for s in shards.values()])

pd.DataFrame(
    {
        name: {"train": len(s["y_train"]), "test": len(s["y_test"]), "mortality": round(float(s["y_train"].mean()), 3)}
        for name, s in shards.items()
    }
).T
```

## Validation AUROC per round

Each client evaluates the incoming global model on its own held-out data before training, so the server log records an out-of-sample curve per site.

```python
import re

log = (WORKSPACE / JOB / "server" / "log.txt").read_text()
per_site: dict[str, list[float]] = {}
for value, site in re.findall(r"validation metric ([0-9.eE+-]+) from client (\S+)", log):
    per_site.setdefault(site, []).append(float(value))

curve = pd.DataFrame(dict(sorted(per_site.items())))
curve.index = range(1, len(curve) + 1)
curve.index.name = "round"

ax = curve.plot(figsize=(7, 4), marker="o", ms=3, alpha=0.7)
curve.mean(axis=1).plot(ax=ax, color="black", lw=2.5, label="mean")
ax.set(xlabel="round", ylabel="AUROC", title="Global model on each site's held-out data")
ax.legend(frameon=False, fontsize=8)
plt.show()
```

## AUROC on the pooled test set

All models share the architecture and training budget; only the data they see differs.
Local models are evaluated on the pooled test set, so each is asked to generalize beyond its own ICU.

```python
models = {f"local {name}": fit(s["X_train"], s["y_train"], EPOCHS) for name, s in shards.items()}
models["federated"] = load_global_model(WORKSPACE, JOB, X_train.shape[1])
models["centralized"] = fit(X_train, y_train, EPOCHS)

table = pd.DataFrame({"auroc": {n: roc_auc_score(y_test, risk_scores(m, X_test)) for n, m in models.items()}})
table.sort_values("auroc").round(3)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(11, 4.5))
for name, model in models.items():
    fpr, tpr, _ = roc_curve(y_test, risk_scores(model, X_test))
    emphasis = name in {"federated", "centralized"}
    axes[0].plot(fpr, tpr, lw=2.2 if emphasis else 1.0, alpha=1.0 if emphasis else 0.5, label=name)
axes[0].plot([0, 1], [0, 1], color="grey", ls=":", lw=1)
axes[0].set(xlabel="false positive rate", ylabel="true positive rate", title="In-hospital mortality, pooled test set")
axes[0].legend(fontsize=8, frameon=False)

ordered = table.sort_values("auroc")
colors = ["tab:grey" if n.startswith("local") else "tab:blue" for n in ordered.index]
axes[1].barh(ordered.index, ordered["auroc"], color=colors)
axes[1].axvline(0.5, color="grey", ls=":", lw=1)
axes[1].set(xlabel="AUROC", xlim=(0.5, min(1.0, ordered["auroc"].max() + 0.05)))
plt.show()
```

## AUROC per site

Each model scored on that site's own test set.

```python
pd.DataFrame(
    {
        site: {
            "own local model": roc_auc_score(s["y_test"], risk_scores(models[f"local {site}"], s["X_test"])),
            "federated": roc_auc_score(s["y_test"], risk_scores(models["federated"], s["X_test"])),
            "centralized": roc_auc_score(s["y_test"], risk_scores(models["centralized"], s["X_test"])),
        }
        for site, s in shards.items()
    }
).T.round(3)
```

## Predicted risk on the embedding

```python
edata = ed.io.read_h5ed(DATA / "hourly.h5ed")
X_all = np.zeros((edata.n_obs, X_train.shape[1]), dtype=np.float32)
for shard in shards.values():
    X_all[shard["rows_train"]] = shard["X_train"]
    X_all[shard["rows_test"]] = shard["X_test"]

edata.obs["predicted_risk"] = risk_scores(models["federated"], X_all)
ep.pl.umap(edata, color=["predicted_risk", "died", "site"], wspace=0.3)
```
