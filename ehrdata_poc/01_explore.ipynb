# PhysioNet 2012 ICU cohort

11988 ICU stays, 37 vitals and labs, 48 hourly intervals, loaded by `ehrdata.dt.physionet2012`.
The outcome is in-hospital mortality.

The object stays three-dimensional throughout.
Where ehrapy requires two-dimensional input it is supplied as a layer or an `.obsm` representation rather than by collapsing the tensor.

Run `python build_dataset.py` before this notebook.

```python
from pathlib import Path

import ehrapy as ep
import ehrdata as ed
import numpy as np

from data import zscore_tensor

DATA = Path("data")
edata = ed.io.read_h5ed(DATA / "hourly.h5ed")
edata
```

## Cohort composition by ICU type

Sites are the unit the patient was admitted to.
They differ in size and in outcome prevalence, which is the non-IID setting federated averaging has to handle.

```python
edata.obs.groupby("site", observed=True).agg(
    patients=("died", "size"),
    mortality=("died", lambda s: (s == "died").mean()),
).round(3)
```

## Missingness and coverage

`ep.pp.qc_metrics` operates on the 3D tensor directly.
Coverage is very uneven: only four variables are observed at more than three timepoints for over 90% of patients, which constrains any trajectory-based distance.

```python
ep.pp.qc_metrics(edata)
observed = (~np.isnan(np.asarray(edata.X, dtype=float))).sum(axis=2)
edata.var["coverage"] = (observed > 3).mean(axis=0)
edata.var[["coverage"]].sort_values("coverage", ascending=False).head(10).round(3)
```

## Differential features by outcome

`ep.tl.rank_features_groups` accepts only 2D input, so it is pointed at the `median` layer holding the per-stay median of each variable.
A 2D layer is valid on a 3D EHRData, so the tensor is unaffected.

```python
ep.tl.rank_features_groups(edata, groupby="died", layer="median")
ep.pl.rank_features_groups(edata, n_features=12)
```

## Embedding from the flattened hourly representation

`ep.pp.neighbors` accepts an `.obsm` representation through `use_rep`, so no collapsing is needed.
`X_flat` is the standardized 1776-dimensional flattening of 37 variables across 48 hours.

```python
ep.pp.neighbors(edata, use_rep="X_flat", n_neighbors=15)
ep.tl.umap(edata)
ep.pl.umap(edata, color=["site", "died", "Age", "SAPS-I"], wspace=0.3)
```

## Embedding from trajectory alignment

`metric="dtw"` compares patients by how their trajectories align rather than by summary values, reading the 3D tensor through `use_rep`.
That key must name a 3D array in `.layers` or `.obsm`; passing `"X"` raises.

Two things matter for this to be meaningful.
Variables must be scaled first, since raw units span three orders of magnitude and the distance would otherwise track whichever high-variance variable happens to be observed.
The distance is quadratic in patients, so this uses a stratified subsample of 500; the full cohort would take roughly a day.

```python
rng = np.random.default_rng(0)
sites = edata.obs["site"].to_numpy()
per_site = [rng.choice(np.flatnonzero(sites == s), 125, replace=False) for s in sorted(set(sites))]
picked = np.sort(np.concatenate(per_site))

sub = edata[picked].copy()
sub.layers["scaled"] = zscore_tensor(edata)[picked]
ep.pp.neighbors(sub, metric="dtw", use_rep="scaled", n_neighbors=15)
ep.tl.umap(sub)
ep.pl.umap(sub, color=["site", "died", "Age"], wspace=0.3)
```

## Persisting the embedding

The EHRData object is written back to `hourly.h5ed`, tensor intact, for `02_federated_results.ipynb`.

```python
ed.io.write_h5ed(edata, DATA / "hourly.h5ed")
```
